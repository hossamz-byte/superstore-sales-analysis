## Handling outliers

info

```python
q1 = df['Sales'].quantile(0.25)
q3 = df['Sales'].quantile(0.75)
iqr = q3 - q1
low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr

outliers = df[(df['Sales'] < low) | (df['Sales'] > high)]
normal = df[(df['Sales'] >= low) & (df['Sales'] <= high)]

print(len(outliers), len(normal), len(df))
```

```python
outliers
```

```python
normal
```

### Outliers
- Used the IQR method on Sales (limits: -272.79 to 500.64).
- Found 1,145 outliers (11.7% of rows), all above the upper limit.
- They make up 64% of total sales and are real large orders (Technology, Furniture, Office Supplies), not errors.
- Decision: kept. Removing them would distort revenue.

# Feature Engineering

```python
df['Shipping Duration'] =(df['Ship Date'] -df['Order Date']).dt.days
```

```python
df
```

```python
df['Sales_Level'] = pd.cut(df['Sales'],bins=[0,df['Sales'].quantile(0.50),q3,df['Sales'].max()],labels=['Low','Medium','High'])
df
```

```python
import numpy as np
df['log_sales'] =np.log(df['Sales']+1)
df
```

```python
df.to_csv(Loc/'data'/ 'Cleaned & featured SuperStore.csv' ,index=False)
```

```python
metric = pd.DataFrame({
    'Mean': [df['Sales'].mean()],
    'Median': [df['Sales'].median()],
    'Skewness': [df['Sales'].skew()],
    'Kurtosis': [df['Sales'].kurtosis()]
}, index=['Sales'])

metric
```

```python
kpis = {
    "Total sales": round(df["Sales"].sum(), 2),
    "Orders": df["Order ID"].nunique(),
    "Customers": df["Customer ID"].nunique(),
    "Avg order value": round(df["Sales"].sum() / df["Order ID"].nunique(), 2),
    "Avg shipping days": round(df["Shipping Duration"].mean(), 2),
    "Top category": df.groupby("Category")["Sales"].sum().idxmax(),
    "Top region": df.groupby("Region")["Sales"].sum().idxmax(),
}

with open(Loc / "data" / "kpis.txt", "w") as f:
    for x, y in kpis.items():
        f.write(f"{x}: {y}\n")
```
